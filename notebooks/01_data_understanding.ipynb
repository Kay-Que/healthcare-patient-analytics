```python
target = 'Test Results'

features = [
    'Age',
    'Gender',
    'Blood Type',
    'Medical Condition',
    'Admission Type',
    'Medication',
    'Billing Amount',
    'Insurance Provider'
]

df[features + [target]].head()
```

```python
X = df[features]
y = df[target]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget classes:")
print(y.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    'Gender',
    'Blood Type',
    'Medical Condition',
    'Admission Type',
    'Medication',
    'Insurance Provider'
]

numeric_features = [
    'Age',
    'Billing Amount'
]

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features),
        ('num', 'passthrough', numeric_features)
    ]
)

X_encoded = preprocessor.fit_transform(X)

print("Encoded feature shape:", X_encoded.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

print("Model training completed successfully.")
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy * 100, 2), "%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression(
    max_iter=1000,
    random_state=42
)

model_lr.fit(X_train, y_train)

y_pred_lr = model_lr.predict(X_test)

accuracy_lr = accuracy_score(y_test, y_pred_lr)

print("Logistic Regression Accuracy:", round(accuracy_lr * 100, 2), "%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_lr))
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier

X_train_dense = X_train.toarray()
X_test_dense = X_test.toarray()

model_gb = HistGradientBoostingClassifier(
    max_iter=200,
    random_state=42
)

model_gb.fit(X_train_dense, y_train)

y_pred_gb = model_gb.predict(X_test_dense)

accuracy_gb = accuracy_score(y_test, y_pred_gb)

print("Gradient Boosting Accuracy:", round(accuracy_gb * 100, 2), "%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_gb))
```

# ML Model Evaluation Summary

Three classification models were tested to predict Test Results: Random Forest, Logistic Regression, and Gradient Boosting.

The models achieved accuracies of 34.25%, 34.20%, and 33.25%, respectively. Since the dataset contains three balanced target classes, the baseline accuracy is approximately 33.3%.

The feature analysis also showed very similar distributions of Test Results across gender, blood type, medical condition, and other categorical variables. This indicates that the available features have limited predictive relationships with Test Results.

Therefore, the models were not considered reliable for practical prediction. The results demonstrate an important finding: not every healthcare dataset contains sufficient predictive patterns for supervised machine learning.
